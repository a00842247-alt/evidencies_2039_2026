## A4: Pivot tables QUIZ



```python
import pandas as pd

df_antrop = pd.read_csv('/antropometria-dataset-2.csv')
df_adult = pd.read_csv('/adult.csv')

for col in df_adult.select_dtypes(include='object').columns:
    df_adult[col] = df_adult[col].str.strip()
```

##Pregunta 1 - Peso promedio en Morelos

```python
pt1 = pd.pivot_table(df_antrop, values='peso', index='desc_ent', aggfunc='mean')
pt1.loc['MORELOS']
```

## Pregunta 2 - Mediana peso sexo 1 en Jalisco

```python
pt2 = pd.pivot_table(df_antrop, values='peso', index='desc_ent', columns='sexo', aggfunc='median', margins=True)
pt2.loc['JALISCO', 1]
```

##Pregunta 3 - Estado con menor peso promedio en sexo 2

```python
pt3 = pd.pivot_table(df_antrop, values='peso', index='desc_ent', columns='sexo', aggfunc='mean')
pt3.sort_values(by=2, ascending=True).head(5)
```

## Pregunta 4 - Promedio age Never-married con Bachelors

```python
pt4 = pd.pivot_table(df_adult, values='age', index=['education', 'marital-status'], aggfunc='mean')
pt4.loc[('Bachelors', 'Never-married')]
```

## Pregunta 5 - Mediana edad Male Married-civ-spouse con 12th

```python
pt5 = pd.pivot_table(
    df_adult,
    values=['age', 'hours-per-week'],
    index=['education', 'marital-status'],
    columns='sex',
    aggfunc='median'
)
pt5.loc[('12th', 'Married-civ-spouse'), ('age', 'Male')]
```

## A5: Visualización con Histogramas & Boxplots

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('antropometria-dataset-1-1.csv')
```

## Pregunta 1 - Box Plot de Cintura

```python
plt.boxplot(df['cintura'].dropna(), vert=False, whis=(0, 100))
plt.title('Box plot de cintura')
plt.show()
```

## Pregunta 2 - Box plot de median PESO de sexo 2

```python
pt = pd.pivot_table(df, values='peso', index='desc_ent', columns='sexo', aggfunc='median')
plt.boxplot(pt[2], vert=False, whis=(0, 100))
plt.title('Box plot median peso sexo 2')
plt.show()
```

## Pregunta 3 - Line plot de cintura de sexo 2

```python
pt3 = pd.pivot_table(df, values='cintura', index='desc_ent', columns='sexo', aggfunc='median')
plt.plot(pt3[2].values)
plt.title('Line plot cintura sexo 2')
plt.show()
```

## Pregunta 4 - Barh plot de ropa de sexo 1

```python
pt4 = pd.pivot_table(df, values='ropa', index='desc_ent', columns='sexo', aggfunc='sum')
top10_ropa = pt4[1].sort_values(ascending=False).head(10)
top10_ropa.plot(kind='barh')
plt.title('Top 10 estados suma de ropa sexo 1')
plt.show()
```

## Pregunta 5 - Histograma de peso

```python
plt.hist(df['peso'].dropna(), bins=25)
plt.title('Histograma de peso')
plt.show()
```
